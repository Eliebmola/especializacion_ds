# 1. Factory Method - Patron Creacional

Tomamos de ejemplo una tienda online en la que se encuentran varios metodos de pagos en el que implementaremos el patron de diseño Factory Method 

1.1 Ejemplo Malo

```python
from abc import ABC, abstractmethod


class TarjetaCredito:
    def pagar(self, monto):
        print(f"Pagando ${monto} con tarjeta de crédito")


class CuentaPayPal:
    def pagar(self, monto):
        print(f"Pagando ${monto} con PayPal")


class ProcesadorPagos:
    def obtener_metodo_pago(self, tipo):
        if tipo == "tarjeta":
            return TarjetaCredito()
        elif tipo == "paypal":
            return CuentaPayPal()
        raise ValueError("Tipo de pago no soportado")
```

```python
procesador = ProcesadorPagos()
procesador.obtener_metodo_pago("tarjeta").pagar(100)
procesador.obtener_metodo_pago("paypal").pagar(57)
```

1.2 Ejemplo Bueno

```python
class MetodoPago(ABC):
    @abstractmethod
    def pagar(self, monto):
        pass


class Tarjeta(MetodoPago):
    def pagar(self, monto):
        print(f"Pagando ${monto} con tarjeta")


class PayPal(MetodoPago):
    def pagar(self, monto):
        print(f"Pagando ${monto} con PayPal")


class FabricaPago(ABC):
    @abstractmethod
    def crear_pago(self) -> MetodoPago:
        pass

    def procesar(self, monto):
        self.crear_pago().pagar(monto)


class FabricaTarjeta(FabricaPago):
    def crear_pago(self):
        return Tarjeta()


class FabricaPayPal(FabricaPago):
    def crear_pago(self):
        return PayPal()
```

```python
FabricaTarjeta().procesar(100)
FabricaPayPal().procesar(59)
```
