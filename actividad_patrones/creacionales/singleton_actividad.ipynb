# 1. Singleton



1.1 Ejemplo bueno

```python
class Config:
    instancia = None

    def __init__(self):
        self.modo_debug = False

    @staticmethod
    def obtener_instancia():
        if Config.instancia is None:
            Config.instancia = Config()
        return Config.instancia


```

```python
config1 = Config.obtener_instancia()
config2 = Config.obtener_instancia()
print("¿Son la misma instancia?", config1 is config2)
```

1.2 Ejemplo malo

```python
class Config_mala:
    def __init__(self):
        self.modo_debug = False
        self.logs = []

    def guardar_log(self, mensaje):
        self.logs.append(mensaje) 
```
