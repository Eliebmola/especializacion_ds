# 2. Factory Method


Tomamos como ejemplo una clase que crea animales


2.1 Ejemplo Bueno


```python
class Perro:
    def hablar(self):
        return "Guau"


class Gato:
    def hablar(self):
        return "Miau"


class CreadorAnimal:
    def crear_animal(self):
        raise NotImplementedError


class CreadorPerro(CreadorAnimal):
    def crear_animal(self):
        return Perro()


class CreadorGato(CreadorAnimal):
    def crear_animal(self):
        return Gato()


def hacer_hablar(creador: CreadorAnimal):
    animal = creador.crear_animal()
    return animal.hablar()
```

```python
print(hacer_hablar(CreadorPerro()))
print(hacer_hablar(CreadorGato()))
```

2.2 Ejemplo malo

```python
def crear_aniimal(tipo):
    if tipo == "perro":
        return Perro()
    elif tipo == "gato":
        return Gato()
    else:
        return None
```

```python
print(crear_aniimal("perro"))
print(crear_aniimal("gato"))
```
