# 5. Prototype

5.1 Ejemplo Bueno

```python
import copy

class  Personaje:
    def __init__(self, nombre, vida, objetos):
        self.nombre = nombre
        self.vida = vida
        self.objetos = objetos

    def clonar(self):
        return copy.deepcopy(self)

    def __str__(self):
        return f"Personaje {self.nombre}, vida={self.vida}, objetos={self.objetos}"
```

```python
personaje_base = Personaje("Guerrero", 100, ["Espafa", "Escudo"])
personaje_clon = personaje_base.clonar()
personaje_clon.vida = 50

print(personaje_base)
print(personaje_clon)
```

5.2 Ejemplo Malo

```python
def clonnar(personaje):
    nuevo = Personaje(personaje.nombre, personaje.vida, personaje.objetos)
    return nuevo
```

```python
personaje_base2 = Personaje("Mago", 80, ["Baston", "Posion"])
personaje_clon2 = clonnar(personaje_base2)
personaje_clon2.objetos.append("Libro hechizos")

print(personaje_base2)
print(personaje_clon2)
```
