# 4. Abstrac Factory


4.1 Ejemplo Bueno

```python
class BotonClaro:
    def dibujar(self):
        return "Boton estilo claro"

class VentanaClara:
    def dibujar(self):
        return "Ventana estilo clara"

class BotonOscuro:
    def dibujar(self):
        return "Boton estilo oscuro"

class VentanaOscura:
    def dibujar(self):
        return "Ventana estilo clara"

```

```python
class FabricarInterfaz:
    def crear_boton(self):
        raise NotImplementedError

    def crear_ventana(self):
        raise NotImplementedError


class FabricaClara(FabricarInterfaz):
    def crear_boton(self):
        return BotonClaro()

    def crear_ventana(self):
        return VentanaClara()

class FabricaOscura(FabricarInterfaz):
    def crear_boton(self):
        return BotonOscuro()

    def crear_ventana(self):
        return VentanaOscura()


def construir_interfaz(fabrica:FabricarInterfaz):
    boton = fabrica.crear_boton()
    ventana = fabrica.crear_ventana()
    return boton.dibujar(), ventana.dibujar()

```

```python
print(construir_interfaz(FabricaClara()))
print(construir_interfaz(FabricaOscura()))
```
