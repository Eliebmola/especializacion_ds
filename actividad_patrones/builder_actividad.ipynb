# 3. Builder

3.1 Ejemplo bueno

```python
class Pizza:
    def __init__(self):
        self.tamano = None
        self.queso = False
        self.pepperoni = False

    def __str__(self):
        return f"Pizza({self.tamano}, queso={self.queso}, pepperoni={self.pepperoni})"
```

```python
class PizzaBuilder:
    def __init__(self):
        self.pizza = Pizza()

    def agregar_tamano(self, tamano):
        self.pizza.tamano = tamano
        return self

    def agregar_queso(self):
        self.pizza.queso = True
        return self

    def agregar_pepperoni(self):
        self.pizza.pepperoni = True
        return self

    def construir(self):
        return self.pizza
```

```python
pizza1 = PizzaBuilder().agregar_tamano("grande").agregar_queso().construir()
print(pizza1)
```
