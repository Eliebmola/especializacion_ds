# 2. Observer - Patron de Comportamiento


Tomamos como ejemplo el envio de notificaciones a distintos medios de comunicación cuando un pedido cambia de estado

2.1 Ejemplo Malo

```python
from abc import ABC, abstractmethod
```

```python
class EmailService:
    def enviar_email(self, msg):
        print(f"EMAIL {msg}")


class SMSService:
    def enviar_sms(self, msg):
        print(f"SMS {msg}")


class Orden:
    def __init__(self):
        self.suscriptores = []

    def suscribir(self, obs):
        self.suscriptores.append(obs)

    def cambiar_estado(self, estado):
        msg = f"Pedido ahora está '{estado}'"
        for obs in self.suscriptores:
            if isinstance(obs, EmailService):
                obs.enviar_email(msg)
            elif isinstance(obs, SMSService):
                obs.enviar_sms(msg)
```

```python
orden = Orden()
orden.suscribir(EmailService())
orden.suscribir(SMSService())
orden.cambiar_estado("enviado")
```

2.2 Ejemplo Bueno

```python
class Observador(ABC):
    @abstractmethod
    def actualizar(self, estado):
        pass


class NotificadorEmail(Observador):
    def actualizar(self, estado):
        print(f"EMAIL Pedido ahora está '{estado}'")


class NotificadorSMS(Observador):
    def actualizar(self, estado):
        print(f"SMS Pedido ahora está '{estado}'")


class Pedido:
    def __init__(self):
        self.suscriptores = []

    def suscribir(self, obs: Observador):
        self.suscriptores.append(obs)

    def cambiar_estado(self, estado):
        for obs in self.suscriptores:
            obs.actualizar(estado)
```

```python
pedido = Pedido()
pedido.suscribir(NotificadorEmail())
pedido.suscribir(NotificadorSMS())
pedido.cambiar_estado("enviado")
```
