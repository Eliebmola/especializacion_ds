# 3. Facade - Patron Estructural


Tomamos como ejemplo el de una tienda online

3.1 Ejemplo Malo

```python
class ProcesadorPago:
    def cobrar(self, monto):
        print(f"Cobrando ${monto}")


class ServicioEnvios:
    def enviar(self, direccion):
        print(f"Enviando pedido a {direccion}")


class GestorCompra:
    def __init__(self):
        self.pago = ProcesadorPago()
        self.envios = ServicioEnvios()

    def cobrar(self, monto):
        self.pago.cobrar(monto)
```

```python
gestor = GestorCompra()
gestor.cobrar(100)
gestor.envios.enviar("Calle 72 #10-15")
```

3.2 Ejemplo Bueno

```python
class PagoService:
    def cobrar(self, monto):
        print(f"Cobrando ${monto}")


class EnvioService:
    def enviar(self, direccion):
        print(f"Enviando pedido a {direccion}")


class CheckoutFacade:
    def __init__(self):
        self._pago = PagoService()
        self._envio = EnvioService()

    def comprar(self, monto, direccion):
        self._pago.cobrar(monto)
        self._envio.enviar(direccion)
```

```python
CheckoutFacade().comprar(100, "Calle 72 #10-15")
```
